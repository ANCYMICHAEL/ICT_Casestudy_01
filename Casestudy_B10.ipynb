# Bank Account Manager

```python
class bank_account:
  def __init__(self,accno,cust_name,balance):
    self.accno = accno
    self.cust_name = cust_name
    self.balance = balance

  def deposit(self, amount):
    if amount>0:
      self.balance+=amount
      print(f"Amount: {amount:.2f} is added to your account. Your A/C balance is {self.balance:.2f}")
    else:
      print("please enter an amount greater than zero")

  def withdraw(self, amount):
    if amount>0:
      if amount>self.balance:
        print("Insufficient balance")
      else:
        self.balance-= amount
        print(f"Amount {amount:.2f} debited from your A/C. Your current balance is {self.balance:.2f}")
    else:
      print("please enter an amount greater than zero")

  def check_balance(self):
    print(f"A/C No: {self.accno} of {self.cust_name} has balance of {self.balance}")

cust1=bank_account(12430100, "Ancy Michael",250000)
cust1.check_balance()
cust1.deposit(5000)
cust1.withdraw(5000)
```

# Weather Data Analyzer


```python
temperature = [30,25,24,21,35]
humidity = [80,85,90,50,74]
wind = [20,25,18,28,30]

avg_temp = sum(temperature)/len(temperature)
avg_humid = sum(humidity)/len(humidity)
avg_wind = sum(wind)/len(wind)

print("Average temperature ",avg_temp)
print("Average humidity ", avg_humid)
print("Average wind ", avg_wind)

if avg_temp>30 and avg_humid<60:
  print("It's Sunny Day")
elif avg_humid>80:
  print("It's Rainy Day")
elif avg_wind>30:
  print("It's Windy Day")
else:
  print("It's Moderate / Cloudy Day")
```
